```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import psycopg2
import os
from dotenv import load_dotenv

load_dotenv()

# Connexion PostgreSQL
conn = psycopg2.connect(
    host=os.getenv("DB_HOST"),
    port=os.getenv("DB_PORT"),
    dbname=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD")
)

# Chargement des données
df_standings = pd.read_sql("SELECT * FROM standings ORDER BY position", conn)
df_matches   = pd.read_sql("SELECT * FROM matches", conn)
df_scorers   = pd.read_sql("SELECT * FROM scorers ORDER BY goals DESC", conn)

print("✅ Données chargées")
print(f"   {len(df_standings)} équipes | {len(df_matches)} matchs | {len(df_scorers)} buteurs")
```

```python
plt.figure(figsize=(12, 6))
colors = ["gold" if i == 0 else "silver" if i == 1 else 
          "chocolate" if i == 2 else "tomato" if i >= 15 else "steelblue" 
          for i in range(len(df_standings))]

bars = plt.barh(df_standings["team"][::-1], df_standings["points"][::-1], color=colors[::-1])
plt.xlabel("Points", fontsize=12)
plt.title("🏆 Classement Ligue 1 — 2025/2026", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("classement.png", dpi=150)
plt.show()
```

```python
goals_per_day = df_matches.groupby("matchday")["total_goals"].mean().reset_index()

plt.figure(figsize=(14, 5))
plt.plot(goals_per_day["matchday"], goals_per_day["total_goals"], 
         marker="o", color="steelblue", linewidth=2)
plt.axhline(y=goals_per_day["total_goals"].mean(), color="red", 
            linestyle="--", label=f"Moyenne : {goals_per_day['total_goals'].mean():.2f}")
plt.xlabel("Journée", fontsize=12)
plt.ylabel("Buts moyens", fontsize=12)
plt.title("⚽ Moyenne de buts par journée — Ligue 1 2025/2026", fontsize=14, fontweight="bold")
plt.legend()
plt.tight_layout()
plt.savefig("buts_par_journee.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_scorers, x="goals", y="player", palette="Reds_r")
plt.xlabel("Buts", fontsize=12)
plt.title("🥇 Top Buteurs Ligue 1 — 2025/2026", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("top_buteurs.png", dpi=150)
plt.show()
```
